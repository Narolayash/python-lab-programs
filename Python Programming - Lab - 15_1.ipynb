### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
class Circle:
    def __init__(self, radious):
        self.r = radious

    def area(self):
        # return 3.14 * (self.r**2)
        print(3.14 * self.r * self.r)

    def perimeter(self):
        print(2 * 3.14 * self.r)

c1 = Circle(5)
c1.area()
c1.perimeter()
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class bank_account:
    def __init__(self, amount):
        self.amount = amount

    def deposit(self,money):
        self.amount += money
        print(money, "added")

    def withdraw(self,money):
        if self.amount - money < 0:
            print("insufficient balance")
            return
        self.amount -= money
        print(money, "withdrawed")

    def checkBalance(self):
        print("balance = ",self.amount)

b1 = bank_account(5000)
b1.deposit(1000)
b1.withdraw(2000)
b1.checkBalance()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, name, price, quantity):
        self.items[name] = [price, quantity]
        print("Item added")

    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
            print("Item removed")
        else:
            print("Item not found")

    def update_item(self, name, price, quantity):
        if name in self.items:
            self.items[name] = [price, quantity]
            print("Item updated")
        else:
            print("Item not found")

    def show_items(self):
        for name in self.items:
            print(name, "Price:", self.items[name][0], "Quantity:", self.items[name][1])
            
inv = Inventory()

inv.add_item("Pen", 10, 50)
inv.add_item("Book", 100, 20)

inv.show_items()

inv.update_item("Pen", 12, 60)
inv.remove_item("Book")

inv.show_items()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute

    def add(self, t1, t2):
        self.minute = t1.minute + t2.minute
        self.hour = t1.hour + t2.hour

        if self.minute >= 60:
            self.hour += 1
            self.minute -= 60

        print("Added Time:", self.hour, "hours", self.minute, "minutes")

t1 = Time(2, 40)
t2 = Time(1, 30)
t3 = Time(0, 0)

t3.add(t1, t2)
```
